```python
# === Chargement du modèle et du processeur ===
model_name = "nateraw/vit-age-classifier"
processor = AutoImageProcessor.from_pretrained(model_name)
model = AutoModelForImageClassification.from_pretrained(model_name)
```

#### Prédiction et évaluation

```python
# === Chargement et prétraitement des images ===
img_paths = [
    Path("C:/Users/remyc/Downloads/visage_perso/25_remy.jpg"),
    Path("C:/Users/remyc/Downloads/visage_perso/44_remy.jpg")
]
images = [Image.open(p).convert("RGB") for p in img_paths]
inputs = processor(images, return_tensors="pt")

# === Prédiction ===
with torch.no_grad():
    outputs = model(**inputs)
    probabilities = torch.nn.functional.softmax(outputs.logits, dim=-1)

# === Affichage des résultats ===
for idx, probs in enumerate(probabilities):
    predicted_class_id = int(torch.argmax(probs).item())
    predicted_label = model.config.id2label[predicted_class_id]
    predicted_confidence = probs[predicted_class_id].item()
    print(f"Image {img_paths[idx].name} : {predicted_label} ({predicted_confidence:.2%} de confiance)")
```
